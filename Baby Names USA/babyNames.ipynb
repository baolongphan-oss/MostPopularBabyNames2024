```python
import pandas as pd
import numpy as np
import glob
```

```python
sorted_txt_files = sorted([filename for filename in glob.glob("names/*.txt")])
all_dfs = []
```

```python
for filename in sorted_txt_files:
    # current_year is formated as "names/yobXXXX.txt", where XXXX is the year
    current_year = int(filename.split("names/yob")[1][:4])
    curr_df = pd.read_csv(filename, sep=',', names=["Name", "Gender", "Number"])
    curr_df["Year"] = current_year
    all_dfs.append(curr_df)
```

```python
df = pd.concat(all_dfs, axis=0).sort_values(by=["Year", "Name", "Gender", "Number"]).reset_index(drop=True)

# Get the first 4 records to view the data model
df.head()
```

```python
# Dataset summary statistics
df.describe(include="all").T
```

```python
# Summary stats for non-numerical columns
df.describe(exclude=np.number).T
```

```python
# Checking for null values
df.isnull().sum()
```

```python
# Checking for duplicate values
df.duplicated(subset=["Name", "Gender"])
```

```python
# View duplicated records
dups = df[df.duplicated(subset=["Name", "Gender"])]
dups.groupby(by="Name")
print(dups["Name"].value_counts())
```

```python
df.to_csv("babyNamesSince1881.csv", index=False)
```
